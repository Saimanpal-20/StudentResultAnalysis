```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Student_score.csv")
print(df.head())
```

```python
df.describe()
```

```python
df.info()
```

```python
df.isnull().sum()
```

Drop unnamed column

```python
df = df.drop("Unnamed: 0", axis = 1)
print(df.head())
```

Change weekly study hours colum

```python
df["WklyStudyHours"] = df["WklyStudyHours"].str.replace("05-Oct" , "5-10")
df.head()
```

check gender distribution

```python
plt.figure(figsize = (5,5))
ax = sns.countplot(data = df, x = "Gender")
ax.bar_label(ax.containers[0])
plt.title("Gender Distribution")
plt.show()
```

From the above chart we have analyseed that:
The number of Female in the data is more than the number of Make

```python
gb = df.groupby("ParentEduc").agg({"MathScore" : 'mean', "ReadingScore" : 'mean' , "WritingScore" : 'mean' })
print(gb)
```

```python
plt.figure(figsize = (5,5))
sns.heatmap(gb, annot = True)
plt.title("Relation ship b/t Parents Education and student Score")
plt.show()
```

#From the above graph we have analysis and concluded that  the education of the parents have a good eimpact on their children education 

```python
gb1 = df.groupby("ParentMaritalStatus").agg({"MathScore" : 'mean', "ReadingScore" : 'mean' , "WritingScore" : 'mean' })
print(gb1)
```

```python
plt.figure(figsize = (5,5))
sns.heatmap(gb1, annot = True)
plt.title("Merital statuts impact on student score")
plt.show()
```

#From the above chart we have concluded that there is no/negligible impact on  the student's score due to their parent's merital ststus

Distribution of Ethnic Groups

```python
groupA = df.loc[df['EthnicGroup']== "group A"].count()
groupB = df.loc[df['EthnicGroup']== "group B"].count()
groupC = df.loc[df['EthnicGroup']== "group C"].count()
groupD = df.loc[df['EthnicGroup']== "group D"].count()
groupE = df.loc[df['EthnicGroup']== "group E"].count()

l = ["group A" , "group B" , "group C" , "group D" , "group E"]
mlist = [groupA["EthnicGroup"] , groupB["EthnicGroup"], groupC["EthnicGroup"], groupD["EthnicGroup"] , groupE["EthnicGroup"]]
plt.pie(mlist , labels = l , autopct = "%1.2f%%")
plt.title("Distribution of Groups")
plt.show()
```
